```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import string
import joblib

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
df = pd.read_csv("/content/sentimentdataset.csv")
```

```python
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print(df.columns)
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
print(df["Sentiment"].value_counts())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Sentiment"
)

plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")

plt.show()
```

```python
def clean_text(text):

    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)

    # Remove HTML tags
    text = re.sub(r"<.*?>", "", text)

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
df["clean_Review"] = df["Text"].apply(clean_text)
```

```python
df[["Text", "clean_Review"]].head()
```

```python
X = df["Text"]
y = df["Sentiment"]
```

```python
print(X.head())
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
vectorizer = TfidfVectorizer(
    max_features=10000,
    ngram_range=(1, 2),
    min_df=2,
    sublinear_tf=True
)
```

```python
X_train_tfidf = vectorizer.fit_transform(X_train)
```

```python
X_test_tfidf = vectorizer.transform(X_test)
```

```python
print("Training shape:", X_train_tfidf.shape)
print("Testing shape:", X_test_tfidf.shape)
```

```python
lr_model = LogisticRegression(
    max_iter=1000
)

lr_model.fit(
    X_train_tfidf,
    y_train
)
```

```python
lr_predictions = lr_model.predict(X_test_tfidf)
```

```python
lr_accuracy = accuracy_score(
    y_test,
    lr_predictions
)

print("Logistic Regression Accuracy:",
      lr_accuracy)
```

```python
print(
    classification_report(
        y_test,
        lr_predictions
    )
)
```

```python
nb_model = MultinomialNB()

nb_model.fit(
    X_train_tfidf,
    y_train
)
```

```python
nb_predictions = nb_model.predict(
    X_test_tfidf
)
```

```python
nb_accuracy = accuracy_score(
    y_test,
    nb_predictions
)

print(
    "Naive Bayes Accuracy:",
    nb_accuracy
)
```

```python
print(
    classification_report(
        y_test,
        nb_predictions
    )
)
```

```python
svm_model = LinearSVC()

svm_model.fit(
    X_train_tfidf,
    y_train
)
```

```python
svm_predictions = svm_model.predict(
    X_test_tfidf
)
```

```python
svm_accuracy = accuracy_score(
    y_test,
    svm_predictions
)

print(
    "SVM Accuracy:",
    svm_accuracy
)
```

```python
print(
    classification_report(
        y_test,
        svm_predictions
    )
)
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Naive Bayes",
        "SVM"
    ],

    "Accuracy": [
        lr_accuracy,
        nb_accuracy,
        svm_accuracy
    ]
})

results
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=results,
    x="Model",
    y="Accuracy"
)

plt.ylim(0, 1)

plt.title("Model Accuracy Comparison")

plt.show()
```

```python
models = {
    "Logistic Regression": lr_model,
    "Naive Bayes": nb_model,
    "SVM": svm_model
}

accuracies = {
    "Logistic Regression": lr_accuracy,
    "Naive Bayes": nb_accuracy,
    "SVM": svm_accuracy
}

best_model_name = max(
    accuracies,
    key=accuracies.get
)

best_model = models[best_model_name]

print("Best Model:", best_model_name)
print("Best Accuracy:", accuracies[best_model_name])
```

```python
best_predictions = best_model.predict(
    X_test_tfidf
)

cm = confusion_matrix(
    y_test,
    best_predictions
)

plt.figure(figsize=(10, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=best_model.classes_,
    yticklabels=best_model.classes_
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")

plt.show()
```

```python
print(
    classification_report(
        y_test,
        best_predictions
    )
)
```

```python
def predict_sentiment(review):

    cleaned = clean_text(review)

    vectorized = vectorizer.transform(
        [cleaned]
    )

    prediction = best_model.predict(
        vectorized
    )[0]

    return prediction
```

```python
review = "I absolutely love this product!"

print(
    predict_sentiment(review)
)
```

```python
review = "This product is terrible and I hate it."

print(
    predict_sentiment(review)
)
```

```python
review = "The product arrived yesterday."

print(
    predict_sentiment(review)
)
```

```python
def predict_with_confidence(review):

    cleaned = clean_text(review)

    vectorized = vectorizer.transform(
        [cleaned]
    )

    prediction = best_model.predict(
        vectorized
    )[0]

    if hasattr(best_model, "predict_proba"):

        probabilities = best_model.predict_proba(
            vectorized
        )[0]

        confidence = max(probabilities) * 100

        return prediction, confidence

    return prediction, None
```
